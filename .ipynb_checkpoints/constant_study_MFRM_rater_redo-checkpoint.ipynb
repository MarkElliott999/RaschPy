# MFRM constant re-derivation: all stages, log-lik AND cos

Re-derives every MFRM smoothing constant under both priority-vector methods
simultaneously, so `(method, constant)` is chosen jointly. See the module
docstring of the generator for full context; headline reasons:

- PCM's same-item-cell matrix bug this session reversed most of that
  model's constant story -- cheap to re-check MFRM's rater PAIR stages
  from scratch.
- Global `matrix_power` default moved 3 -> 5; the cos arm sweeps power
  {3,5,8} to confirm the constant re-tune doesn't shift that.
- Whether MFRM's default `method` should switch log-lik -> cos is open;
  the constants were all originally swept under log-lik, and this removes
  that caveat.

`person_sd` pegged to combined item+threshold range; old fixed 1.5 kept as
a narrow-SD edge case.

```
import sys, time, warnings, itertools
from pathlib import Path
import numpy as np, pandas as pd
warnings.filterwarnings("ignore")
import raschpy as rp
print("raschpy:", rp.__file__)

OUT = Path("constant_study_outputs"); OUT.mkdir(exist_ok=True)
RUN_TAG = "rater_redo"
```

## Helpers, grids, method configs

```
def pegged_sd(item_range, K, category_base=1.0):
    return max(0.5, (item_range + (K - 1) * category_base) / 4)

NARROW_SD = 1.5

def resolve_sd(item_range, K, mode, category_base=1.0):
    return pegged_sd(item_range, K, category_base) if mode == "pegged" else NARROW_SD

def rec(gen, est):
    x, y = np.asarray(gen, float).ravel(), np.asarray(est, float).ravel()
    if x.shape != y.shape:
        return (np.nan, np.nan)
    ok = np.isfinite(x) & np.isfinite(y)
    x, y = x[ok], y[ok]
    if len(x) < 2:
        return (np.nan, np.nan)
    xc, yc = x - x.mean(), y - y.mean()
    ss = float(xc @ xc)
    rmse = float(np.sqrt(np.mean((xc - yc) ** 2)))
    beta = float(xc @ yc / ss) if (xc.std() >= 1e-6 and ss > 0) else np.nan
    return (rmse, beta)

CS = [0.0, 0.02, 0.05, 0.1, 0.2, 0.3, 0.63, 1.0, 1.5, 2.5, 4.0]
REPS = 20
SEED0 = 41000
FLOOR = 0.85

# (label, method, matrix_power) -- matrix_power None lets calibrate resolve it
# (0 for log-lik, 5 for cos post the base.py change). The cos rows pin 3/5/8
# explicitly to confirm the constant re-tune doesn't move the power optimum.
METHOD_CONFIGS = [
    ("loglik",  "log-lik", None),
    ("cos_p3",  "cos",     3),
    ("cos_p5",  "cos",     5),
    ("cos_p8",  "cos",     8),
]
```

## Condition grid

Core I x K spans the I/K density axis (the mechanism behind the
items-vs-thresholds asymmetry in the first MFRM redo). One-factor sweeps on
R, N, missing, item_range, facet_range. Stretch arms add a stretch_range
sweep (the spread of the true omega_r itself).

```
BASE = dict(I=8, K=6, R=8, N=200, item_range=2.0, facet_range=2.0,
            stretch_range=1.0, missing=0.0, category_base=1.0)

def cond(label, sd_mode="pegged", **ov):
    d = dict(BASE); d.update(ov); d["label"] = label; d["sd_mode"] = sd_mode
    return d

CONDS = []
for I, K in [(4, 3), (8, 6), (16, 12), (8, 3), (16, 3)]:
    CONDS.append(cond(f"core_I{I}_K{K}", I=I, K=K))
CONDS.append(cond("core_I8_K6_narrow", sd_mode="narrow"))
for R in [4, 16]:
    CONDS.append(cond(f"R{R}", R=R))
for N in [100, 400]:
    CONDS.append(cond(f"N{N}", N=N))
for miss in [0.3, 0.6]:
    CONDS.append(cond(f"miss{miss}", missing=miss))
for ir in [1.0, 4.0]:
    CONDS.append(cond(f"itemrange{ir}", item_range=ir))
for fr in [1.0, 4.0]:
    CONDS.append(cond(f"facetrange{fr}", facet_range=fr))

STRETCH_EXTRA = [cond(f"stretchrange{sr}", stretch_range=sr) for sr in [0.5, 2.0]]

print(len(CONDS), "shared conditions;", len(STRETCH_EXTRA), "extra for stretch arms")
```

## Shared per-rep driver

`build_model(cond, seed, simcls)` -> (sim, m) with the item/threshold
LOCATIONS overwritten to their true generating values, so each rater arm's
constant is measured against a clean location input. `est_kwargs(mcfg, c)`
assembles the estimator call.

```
def build_model(c_, seed, simcls, stretch=False):
    # MFRM_Sim_{Items,Thresholds,Matrix,Bivector} take facet_range via **kw;
    # MFRM_Sim_{Centrality,PseudoHalo,Bistretch} reject facet_range and use
    # `global_range` (spread of true lambda_r) + `stretch_range` (spread of
    # true omega_r) instead, forwarding item_range/person_sd/missing/seed.
    I, K, R, N = c_["I"], c_["K"], c_["R"], c_["N"]
    sd = resolve_sd(c_["item_range"], K, c_["sd_mode"], c_["category_base"])
    if stretch:
        kw = dict(item_range=c_["item_range"], global_range=c_["facet_range"],
                  stretch_range=c_["stretch_range"], person_sd=sd,
                  missing=c_["missing"], seed=seed)
    else:
        kw = dict(item_range=c_["item_range"], facet_range=c_["facet_range"],
                  person_sd=sd, missing=c_["missing"], seed=seed)
    sim = simcls(I, N, K, R, **kw)
    m = rp.MFRM(sim.responses, validate=False)
    m.items = pd.Series(np.asarray(sim.items, float), index=sim.item_names)
    m.thresholds = pd.Series(np.asarray(sim.thresholds, float).ravel(),
                             index=range(1, K + 1))
    return sim, m, sd

def est_kwargs(mcfg, c):
    _, method, mp = mcfg
    kw = dict(constant=c, method=method)
    if mp is not None:
        kw["matrix_power"] = mp
    return kw
```

## Arm L: item-location + threshold-location constants

`item_diffs` uses the priority-vector method (swept over all 4 configs);
`ra_thresholds` is CPAT (method-independent, swept once). Both measured
against true generating values.

```
rows_loc, t0 = [], time.time()
seed = SEED0
for ci, c_ in enumerate(CONDS):
    I, K, R, N = c_["I"], c_["K"], c_["R"], c_["N"]
    for rep in range(REPS):
        seed += 1
        try:
            sim = rp.MFRM_Sim(I, N, K, R, item_range=c_["item_range"],
                              facet_range=c_["facet_range"],
                              person_sd=resolve_sd(c_["item_range"], K, c_["sd_mode"], c_["category_base"]),
                              missing=c_["missing"], seed=seed)
        except Exception:
            continue
        # threshold-location sweep (method-independent): needs true item locs
        true_items = pd.Series(np.asarray(sim.items, float), index=sim.item_names)
        for c in CS:
            try:
                m = rp.MFRM(sim.responses, validate=False)
                thr = m.ra_thresholds(true_items, constant=c)
                tr, tb = rec(sim.thresholds, thr)
                rows_loc.append(dict(c_, stage="thr_loc", mcfg="cpat", constant=c,
                                     rmse=tr, beta=tb))
            except Exception:
                pass
        # item-location sweep (method-dependent)
        for mcfg in METHOD_CONFIGS:
            for c in CS:
                try:
                    m = rp.MFRM(sim.responses, validate=False)
                    m.item_diffs(**est_kwargs(mcfg, c))
                    ir, ib = rec(sim.items, m.items)
                    rows_loc.append(dict(c_, stage="item_loc", mcfg=mcfg[0], constant=c,
                                         rmse=ir, beta=ib))
                except Exception:
                    pass
    if (ci + 1) % 4 == 0 or ci == len(CONDS) - 1:
        el = (time.time() - t0) / 60
        print(f"  L: cond {ci+1}/{len(CONDS)}  {el:.1f}m", flush=True)

df_loc = pd.DataFrame(rows_loc)
df_loc.to_csv(OUT / f"MFRM_location_{RUN_TAG}.csv", index=False)
print("done,", len(df_loc), "rows")
```

## Arms R1-R4: rater-facet vectors judged on RMSE-vs-beta balance

items / thresholds / matrix / bivector. `facet_effects_*` vs the sim's true
`facet_effects`.

```
RATER_BALANCE = [
    ("items",      rp.MFRM_Sim_Items,      "_estimate_raters_items",      "facet_effects_items"),
    ("thresholds", rp.MFRM_Sim_Thresholds, "_estimate_raters_thresholds", "facet_effects_thresholds"),
    ("matrix",     rp.MFRM_Sim_Matrix,     "_estimate_raters_matrix",     "facet_effects_matrix"),
    ("bivector",   rp.MFRM_Sim_Bivector,   "_estimate_raters_bivector",   "facet_effects_bivector"),
]

for arm, simcls, estname, outattr in RATER_BALANCE:
    rows, t0 = [], time.time()
    seed = SEED0 + 100000 * (RATER_BALANCE.index((arm, simcls, estname, outattr)) + 1)
    for ci, c_ in enumerate(CONDS):
        for rep in range(REPS):
            seed += 1
            try:
                sim, m, sd = build_model(c_, seed, simcls)
            except Exception:
                continue
            true_fe = np.asarray(sim.facet_effects.values, float).ravel()
            for mcfg in METHOD_CONFIGS:
                for c in CS:
                    try:
                        getattr(m, estname)(**est_kwargs(mcfg, c))
                        est = getattr(m, outattr)
                        est = np.asarray(est.values if hasattr(est, "values") else est, float).ravel()
                        r, b = rec(true_fe, est)
                        rows.append(dict(c_, mcfg=mcfg[0], constant=c, rmse=r, beta=b, person_sd=sd))
                    except Exception:
                        pass
        if (ci + 1) % 6 == 0 or ci == len(CONDS) - 1:
            el = (time.time() - t0) / 60
            print(f"  {arm}: cond {ci+1}/{len(CONDS)}  {el:.1f}m", flush=True)
    df = pd.DataFrame(rows)
    df.to_csv(OUT / f"MFRM_{arm}_{RUN_TAG}.csv", index=False)
    print(f"  {arm} done, {len(df)} rows")
```

## Arms R5-R7: stretch models judged on omega_r recovery

centrality / pseudo_halo (omega_r + lambda_r) and bistretch (omega_items_r +
omega_thresholds_r + lambda_r). omega_r's own recovery is the target;
lambda_r reported alongside (pseudo_halo has a known lambda-vs-omega
conflict). Stretch-range sweep added.

```
CENT_CONDS = CONDS + STRETCH_EXTRA

def run_stretch(arm, simcls, estname, truth_map, est_maps, seed_base):
    rows, t0 = [], time.time()
    seed = seed_base
    for ci, c_ in enumerate(CENT_CONDS):
        for rep in range(REPS):
            seed += 1
            try:
                sim, m, sd = build_model(c_, seed, simcls, stretch=True)
            except Exception:
                continue
            truth = {k: np.asarray(getattr(sim, v), float).ravel() for k, v in truth_map.items()}
            for mcfg in METHOD_CONFIGS:
                for c in CS:
                    try:
                        kw = est_kwargs(mcfg, c); kw["regression"] = "huber"
                        getattr(m, estname)(**kw)
                        row = dict(c_, mcfg=mcfg[0], constant=c, person_sd=sd)
                        for k, attr in est_maps.items():
                            est = np.asarray(getattr(m, attr), float).ravel()
                            r, b = rec(truth[k], est)
                            row[f"{k}_rmse"], row[f"{k}_beta"] = r, b
                        rows.append(row)
                    except Exception:
                        pass
        if (ci + 1) % 6 == 0 or ci == len(CENT_CONDS) - 1:
            el = (time.time() - t0) / 60
            print(f"  {arm}: cond {ci+1}/{len(CENT_CONDS)}  {el:.1f}m", flush=True)
    df = pd.DataFrame(rows)
    df.to_csv(OUT / f"MFRM_{arm}_{RUN_TAG}.csv", index=False)
    print(f"  {arm} done, {len(df)} rows")
    return df

df_cent = run_stretch("centrality", rp.MFRM_Sim_Centrality, "_estimate_raters_centrality",
                      {"omega": "omega", "lam": "lambda_"},
                      {"omega": "omega_centrality", "lam": "lambda_centrality"},
                      SEED0 + 500000)
df_ph = run_stretch("pseudo_halo", rp.MFRM_Sim_PseudoHalo, "_estimate_raters_pseudo_halo",
                    {"omega": "omega", "lam": "lambda_"},
                    {"omega": "omega_pseudo_halo", "lam": "lambda_pseudo_halo"},
                    SEED0 + 600000)
df_bis = run_stretch("bistretch", rp.MFRM_Sim_Bistretch, "_estimate_raters_bistretch",
                     {"omega_it": "omega_items", "omega_th": "omega_thresholds", "lam": "lambda_"},
                     {"omega_it": "omega_items_bistretch", "omega_th": "omega_thresholds_bistretch",
                      "lam": "lambda_bistretch"},
                     SEED0 + 700000)
```

## Results

```
def balance_report(df, rmse_col="rmse", beta_col="beta", by=("label", "mcfg")):
    for key, g in df.groupby(list(by)):
        t = g.groupby("constant").agg(rmse=(rmse_col, "mean"), beta=(beta_col, "mean"))
        if t["rmse"].isna().all():
            print(f"  {str(key):40} -- all NaN"); continue
        elig = t[t.beta >= FLOOR]
        cstar = elig["rmse"].idxmin() if len(elig) else (t["beta"].idxmax() if t["beta"].notna().any() else t["rmse"].idxmin())
        print(f"  {str(key):40} c*={cstar:<6g} rmse={t.loc[cstar,'rmse']:.4f} beta={t.loc[cstar,'beta']:.3f}")

def method_summary(df, rmse_col="rmse", beta_col="beta", label="", core_only=True):
    d = df[df.label.str.startswith("core_")] if core_only else df
    print(f"--- {label} (core conditions, pooled) ---")
    for mcfg in ["loglik", "cos_p3", "cos_p5", "cos_p8"]:
        g = d[d.mcfg == mcfg]
        if not len(g):
            print(f"  {mcfg:8}  (no rows)"); continue
        t = g.groupby("constant").agg(rmse=(rmse_col, "mean"), beta=(beta_col, "mean"))
        elig = t[t.beta >= FLOOR]
        cstar = elig["rmse"].idxmin() if len(elig) else t["beta"].idxmax()
        curve = " ".join(f"{c:g}:{t.loc[c,'rmse']:.3f}/{t.loc[c,'beta']:.2f}" for c in CS if c in t.index)
        print(f"  {mcfg:8}  c*={cstar:<6g} rmse={t.loc[cstar,'rmse']:.4f} beta={t.loc[cstar,'beta']:.3f}   {curve}")

df_loc = pd.read_csv(OUT / f"MFRM_location_{RUN_TAG}.csv")
print("=== ITEM LOCATION (item_diffs) ===")
method_summary(df_loc[df_loc.stage == "item_loc"], label="item location")
print("\n=== THRESHOLD LOCATION (ra_thresholds / CPAT, method-independent) ===")
tl = df_loc[df_loc.stage == "thr_loc"]
tlc = tl[tl.label.str.startswith("core_")]
t = tlc.groupby("constant").agg(rmse=("rmse", "mean"), beta=("beta", "mean"))
elig = t[t.beta >= FLOOR]
cstar = elig["rmse"].idxmin() if len(elig) else t["beta"].idxmax()
print(f"  c*={cstar:<6g} rmse={t.loc[cstar,'rmse']:.4f} beta={t.loc[cstar,'beta']:.3f}   " +
      " ".join(f"{c:g}:{t.loc[c,'rmse']:.3f}/{t.loc[c,'beta']:.2f}" for c in CS if c in t.index))
```

```
for arm in ["items", "thresholds", "matrix", "bivector"]:
    df = pd.read_csv(OUT / f"MFRM_{arm}_{RUN_TAG}.csv")
    method_summary(df, label=f"{arm} rater-facet")
    print()
```

```
def omega_summary(df, omega_cols, lam_col="lam", label=""):
    d = df[df.label.str.startswith("core_")]
    print(f"--- {label} (core, pooled) ---")
    for mcfg in ["loglik", "cos_p3", "cos_p5", "cos_p8"]:
        g = d[d.mcfg == mcfg]
        if not len(g):
            print(f"  {mcfg:8}  (no rows)"); continue
        aggd = {f"{oc}_rmse": (f"{oc}_rmse", "mean") for oc in omega_cols}
        aggd.update({f"{oc}_beta": (f"{oc}_beta", "mean") for oc in omega_cols})
        aggd[f"{lam_col}_rmse"] = (f"{lam_col}_rmse", "mean")
        aggd[f"{lam_col}_beta"] = (f"{lam_col}_beta", "mean")
        t = g.groupby("constant").agg(**aggd)
        oc0 = omega_cols[0]
        if t[f"{oc0}_beta"].isna().all():
            print(f"  {mcfg:8}  all NaN"); continue
        cstar = (t[f"{oc0}_beta"] - 1).abs().idxmin()
        parts = []
        for oc in omega_cols:
            parts.append(f"{oc} rmse={t.loc[cstar, oc+'_rmse']:.4f}/beta={t.loc[cstar, oc+'_beta']:.3f}")
        print(f"  {mcfg:8}  c*(|omega_beta-1| min)={cstar:<6g}  " + "  ".join(parts) +
              f"   lam_beta_there={t.loc[cstar, lam_col+'_beta']:.3f}")

def _load(name):
    fp = OUT / f"MFRM_{name}_{RUN_TAG}.csv"
    try:
        d = pd.read_csv(fp)
        return d if len(d) else None
    except Exception:
        return None

df_cent, df_ph, df_bis = _load("centrality"), _load("pseudo_halo"), _load("bistretch")
if df_cent is not None:
    omega_summary(df_cent, ["omega"], label="centrality -- omega_r recovery")
    print()
else:
    print("centrality: no rows -- structural failure, inspect the arm")
if df_ph is not None:
    omega_summary(df_ph, ["omega"], label="pseudo_halo -- omega_r recovery (watch lambda_r)")
    print()
else:
    print("pseudo_halo: no rows")
if df_bis is not None:
    omega_summary(df_bis, ["omega_it", "omega_th"], label="bistretch -- omega_items_r + omega_thresholds_r")
else:
    print("bistretch: no rows")
```

## Notes

- **Method question:** for each stage, compare `loglik` vs `cos_p5` at each
  one's own c*. If accuracy is tied (as it was at fixed c=0.1), cos is the
  better default (faster, more robust on thin rater graphs). If loglik has
  a real edge anywhere once the constant is re-tuned, that stage argues
  against the global switch.
- **Power question:** cos_p3 vs cos_p5 vs cos_p8 -- if p5 == p8 and both
  beat p3, the base.py power-5 default holds after the constant re-tune. If
  p3 == p5, the change is inert for MFRM (as the fixed-c sweep suggested).
- **Constant portability:** if c* is the same for loglik and every cos row
  in a stage, the constant ports and only the method/power choice is open.
  A stage where cos wants a different c* than loglik needs its own value.
- **items/thresholds:** the shipped `robust=True` facet default is 0.63.
  Check whether that still lands under the balance criterion here, and
  whether the well-conditioned (`robust=False`) default should stay 0.1.
- **matrix/bivector:** currently 0.3. Confirm or revise.
- **centrality:** was 0.1, confirmed low. **pseudo_halo:** the lambda-vs-
  omega conflict -- does it survive, and does cos change its shape?
  **bistretch:** never re-checked against the I/K-density mechanism or the
  omega-primacy criterion before -- this is its first proper sweep.

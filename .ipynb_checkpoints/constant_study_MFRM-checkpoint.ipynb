# MFRM `constant` calibration study

Systematic simulation to choose the default `constant` for the three MFRM calibration stages under `method='log-lik'`:

* **item** stage (`item_diffs`, PAIR) - model-independent
* **threshold** stage (`ra_thresholds`, CPAT) - model-independent
* **rater** stage (`_estimate_raters_*`) - representation-specific; `global` and `matrix` are swept (the `matrix` marginal drives `bivector` / `bistretch`)

Each replication is calibrated once at `REF_CONSTANT`; each stage is then re-run across its own constant grid with the other stages held at the reference, so the stages are isolated. `CONSTANTS_RATER` is dense through the `rater_matrix` bias/RMSE tension zone - see the sweet-spot section.

````python
import os, sys, time, warnings, itertools
from collections import defaultdict
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")
sys.path.insert(0, str(Path.cwd()))
import raschpy as rp

print("raschpy from:", rp.__file__)
import inspect
_dflt = inspect.signature(rp.SLM.calibrate).parameters["method"].default
assert _dflt == "log-lik", f"expected log-lik default, got {_dflt!r} - wrong raschpy on path"
OUTDIR = Path("constant_study_outputs"); OUTDIR.mkdir(exist_ok=True)
print("outputs ->", OUTDIR.resolve())
````

````python
def recovery_metrics(est, gen):
    """Recovery of `est` against generating `gen`, on mean-centred vectors.

    beta      OLS slope of estimate on truth   (<1 compresses toward the mean)
    sd_ratio  sd(est) / sd(truth)              (<1 compresses)
    rmse      sqrt(mean (est-truth)^2)         Luecht & Ackerman (2018) residual
    mad       mean |est-truth|
    r         Pearson correlation
    ok        1 if finite and shape-matched, else 0

    beta / sd_ratio / r need a vector of >= 3 elements with a non-degenerate
    generating spread; on a 2-element vector (e.g. RSM thresholds at max_score=2)
    the "slope" is just a volatile ratio, so they are NaN while rmse / mad stand.
    """
    e = np.asarray(est, dtype=float).ravel()
    g = np.asarray(gen, dtype=float).ravel()
    if e.shape != g.shape or e.size < 2 or not np.isfinite(e).all():
        return dict(beta=np.nan, sd_ratio=np.nan, rmse=np.nan, mad=np.nan, r=np.nan, ok=0)
    ec = e - e.mean()
    gc = g - g.mean()
    ss = float(np.dot(gc, gc))
    if ss == 0:
        return dict(beta=np.nan, sd_ratio=np.nan, rmse=np.nan, mad=np.nan, r=np.nan, ok=0)
    resid = ec - gc
    out = dict(rmse=float(np.sqrt(np.mean(resid ** 2))),
               mad=float(np.mean(np.abs(resid))), ok=1)
    if e.size >= 3 and gc.std() >= 0.05:
        out.update(beta=float(np.dot(gc, ec) / ss),
                   sd_ratio=float(ec.std() / gc.std()),
                   r=float(np.corrcoef(ec, gc)[0, 1]))
    else:
        out.update(beta=np.nan, sd_ratio=np.nan, r=np.nan)
    return out

FAIL = dict(beta=np.nan, sd_ratio=np.nan, rmse=np.nan, mad=np.nan, r=np.nan, ok=0)
METRIC_COLS = ["beta", "sd_ratio", "rmse", "mad", "r"]
ROBUST_COLS = ["beta", "sd_ratio", "r"]   # median across conditions
MEAN_COLS = ["rmse", "mad"]               # mean across conditions

def condition_rows(base, rep_metrics):
    """rep_metrics: dict[(vector, constant)] -> list of per-rep metric dicts.
    Returns one mean-over-reps row per (vector, constant); raw rep rows are
    discarded so memory stays O(conditions), not O(conditions x reps)."""
    out = []
    for (vec, c), lst in rep_metrics.items():
        d = pd.DataFrame(lst)
        row = dict(base)
        row.update(vector=vec, constant=c, reps=len(d), n_ok=int(d["ok"].sum()))
        for mcol in METRIC_COLS:
            row[mcol] = float(d[mcol].mean())   # skips NaN
        out.append(row)
    return out
````

## Configuration

Edit the grids below. `QUICK = True` runs a small smoke test; `QUICK = False` runs the full overnight study. Seeds run from `SEED_START` (default 1), incrementing by 1 for every simulated dataset, so a given config is exactly reproducible.

````python
QUICK = False
SEED_START = 1

CONSTANTS_ITEM   = [0.0, 0.03, 0.06, 0.09, 0.13, 0.18, 0.25, 0.35, 0.5, 0.7, 1.0]
CONSTANTS_THRESH = [0.02, 0.035, 0.05, 0.07, 0.1, 0.14, 0.19, 0.26, 0.36, 0.5]
CONSTANTS_RATER  = [0.05, 0.1, 0.15, 0.2, 0.28, 0.36, 0.45, 0.55, 0.7, 0.85, 1.0,
                    1.2, 1.45, 1.75, 2.1, 2.5, 3.0, 3.6, 4.4, 5.5, 7.0]
REF_CONSTANT     = 0.3

N_ITEMS     = [6, 9, 12, 20]
N_PERSONS   = [50, 100, 150, 500, 1000]
N_RATERS    = [3, 6, 9, 15]
MAX_SCORE   = [3, 5, 7]
ITEM_RANGE  = [3.0, 5.0]         # generating item-difficulty spread
FACET_RANGE = [1.0, 2.0, 3.5]    # generating rater-effect spread
PERSON_SD   = [1.2, 2.0]
MISSING     = [0.0, 0.25, 0.5, 0.8]
REPS        = 50
# ~1300 conditions x 30 reps x 63 constant-evals. Rough wall time 6-9 h (the
# large n_items x n_raters x n_persons cells dominate). To trim: drop REPS,
# drop a N_RATERS / N_PERSONS level, or shorten CONSTANTS_RATER's tail.

if QUICK:
    CONSTANTS_ITEM   = [0.0, 0.09, 0.25, 0.5]
    CONSTANTS_THRESH = [0.035, 0.1, 0.19, 0.36]
    CONSTANTS_RATER  = [0.1, 0.36, 0.7, 1.2, 2.1, 3.6, 7.0]
    N_ITEMS, N_PERSONS, N_RATERS = [10], [300], [4, 12]
    MAX_SCORE, ITEM_RANGE, FACET_RANGE = [4], [4.0], [2.0]
    PERSON_SD, MISSING = [1.5], [0.0, 0.4]
    REPS = 4
````

````python
_SKIP = METRIC_COLS + ["vector", "constant", "n_ok", "reps", "rep"]

def _agg(df, keys):
    """Across conditions: median of slope/spread metrics (robust to the few
    near-degenerate conditions), mean of the residual metrics."""
    d = df.copy()
    d.loc[d["beta"].abs() > 3, ROBUST_COLS] = np.nan
    g = d.groupby(list(keys))
    return pd.concat([g[ROBUST_COLS].median(), g[MEAN_COLS].mean()], axis=1).reset_index()

def by_constant(cond_df, extra_keys=()):
    return _agg(cond_df, ["vector", "constant", *extra_keys])

def show_marginal(cond_df):
    for vec, g in cond_df.groupby("vector"):
        marg = _agg(g, ["constant"]).set_index("constant")[METRIC_COLS].round(4)
        print(f"\n=== {vec}: by constant  (beta/sd_ratio/r = median, rmse/mad = mean) ===")
        print(marg.to_string())

def recommend(cond_df, tols=(0.02, 0.05, 0.10)):
    """Per vector: rmse-optimal, bias-constrained-optimal at several |beta-1|
    tolerances, and the minimax (worst per-condition RMSE ratio) constant."""
    rows = []
    for vec, g in cond_df.groupby("vector"):
        marg = _agg(g, ["constant"]).set_index("constant").sort_index()
        marg = marg[np.isfinite(marg["rmse"])]
        if marg.empty:
            continue
        ab = (marg["beta"] - 1).abs()
        rec = dict(vector=vec,
                   rmse_opt=marg["rmse"].idxmin(),
                   beta_at_rmse_opt=round(float(marg["beta"].loc[marg["rmse"].idxmin()]), 3))
        for t in tols:
            near = marg[ab <= t]
            rec[f"bias_opt@{t:g}"] = (near["rmse"].idxmin() if not near.empty
                                      else ab.idxmin())
        piv = g.pivot_table(index=[c for c in g.columns if c not in _SKIP],
                            columns="constant", values="rmse")
        ratio = piv.div(piv.min(axis=1), axis=0)
        rec["minimax"] = ratio.max(axis=0).idxmin()
        rec["worst_ratio_minimax"] = round(float(ratio.max(axis=0).min()), 3)
        rows.append(rec)
    return pd.DataFrame(rows)

def tradeoff(cond_df, vec, bias_weight=1.0):
    """Bias vs RMSE by constant for one vector, with Pareto flag and a
    scalarised score  = (rmse/min_rmse - 1) + bias_weight * |beta - 1|."""
    m = _agg(cond_df[cond_df["vector"] == vec], ["constant"]).set_index("constant").sort_index()
    m = m[np.isfinite(m["rmse"])].copy()
    m["abs_bias"] = (m["beta"] - 1).abs()
    m["rmse_rel"] = m["rmse"] / m["rmse"].min()
    dom = []
    for _, row in m.iterrows():
        d = (((m["abs_bias"] <= row["abs_bias"]) & (m["rmse"] <= row["rmse"]) &
              ((m["abs_bias"] < row["abs_bias"]) | (m["rmse"] < row["rmse"]))).any())
        dom.append(not d)
    m["pareto"] = dom
    m["score"] = (m["rmse_rel"] - 1.0) + bias_weight * m["abs_bias"].fillna(np.inf)
    return m[["beta", "abs_bias", "sd_ratio", "rmse", "rmse_rel", "pareto", "score"]].round(4)

def sweet_spots(cond_df, weights=(0.5, 1.0, 2.0)):
    rows = []
    for vec in sorted(cond_df["vector"].unique()):
        r = dict(vector=vec)
        for w in weights:
            t = tradeoff(cond_df, vec, bias_weight=w)
            r[f"w={w:g}"] = t["score"].idxmin() if t["score"].notna().any() else np.nan
        rows.append(r)
    return pd.DataFrame(rows)
````

````python
def plot_vector(cond_df, vec, facet="n_items", fname=None):
    g = cond_df[cond_df["vector"] == vec]
    levels = sorted(g[facet].unique())
    fig, ax = plt.subplots(1, 3, figsize=(15, 4))
    for lv in levels:
        s = g[g[facet] == lv].groupby("constant")[["rmse", "beta", "sd_ratio"]].mean()
        ax[0].plot(s.index, s["rmse"], marker="o", ms=3, label=f"{facet}={lv}")
        ax[1].plot(s.index, s["beta"], marker="o", ms=3, label=f"{facet}={lv}")
        ax[2].plot(s.index, s["sd_ratio"], marker="o", ms=3, label=f"{facet}={lv}")
    ax[0].set(title=f"{vec}: RMSE vs constant", xlabel="constant", ylabel="RMSE")
    ax[1].axhline(1.0, color="k", lw=0.8, ls="--")
    ax[1].set(title=f"{vec}: slope beta vs constant", xlabel="constant", ylabel="beta")
    ax[2].axhline(1.0, color="k", lw=0.8, ls="--")
    ax[2].set(title=f"{vec}: sd-ratio vs constant", xlabel="constant", ylabel="sd(est)/sd(true)")
    for a in ax:
        a.legend(fontsize=7); a.grid(alpha=0.3)
    fig.tight_layout()
    if fname:
        fig.savefig(OUTDIR / fname, dpi=110)
    plt.show()

def plot_tradeoff(cond_df, vec, fname=None):
    t = tradeoff(cond_df, vec)
    fig, axx = plt.subplots(figsize=(6, 4.5))
    axx.plot(t["abs_bias"], t["rmse"], "-", color="0.7", lw=1, zorder=1)
    axx.scatter(t["abs_bias"], t["rmse"], c=t.index, cmap="viridis", zorder=2)
    pf = t[t["pareto"]]
    axx.scatter(pf["abs_bias"], pf["rmse"], s=120, facecolors="none",
                edgecolors="red", label="Pareto", zorder=3)
    for c, row in t.iterrows():
        axx.annotate(f"{c:g}", (row["abs_bias"], row["rmse"]), fontsize=6)
    axx.set(title=f"{vec}: bias-RMSE tradeoff (label = constant)",
            xlabel="|beta - 1|  (compression)", ylabel="RMSE")
    axx.legend(); axx.grid(alpha=0.3)
    fig.tight_layout()
    if fname:
        fig.savefig(OUTDIR / fname, dpi=110)
    plt.show()
````

````python
def _eta(done, total, t0):
    el = time.time() - t0
    return f"{done}/{total}  elapsed {el/60:.1f}m  eta {el/max(done,1)*(total-done)/60:.1f}m"
````

## Run the simulation sweep

Streams condition-level rows into `cond` (one mean-over-reps row per `vector` x `constant` x condition); raw per-rep rows are discarded as it goes, so memory stays flat regardless of grid size.

````python
grid = list(itertools.product(N_ITEMS, N_PERSONS, N_RATERS, MAX_SCORE,
                              ITEM_RANGE, FACET_RANGE, PERSON_SD, MISSING))
n_ce = len(CONSTANTS_ITEM) + len(CONSTANTS_THRESH) + 2 * len(CONSTANTS_RATER)
print(len(grid), "conditions x", REPS, "reps x", n_ce, "constant-evals;",
      len(grid) * REPS * 2, "sims")
seeds = itertools.count(SEED_START)
crows, t0 = [], time.time()

for ci, (n_i, n_p, n_r, m_s, irange, frange, psd, miss) in enumerate(grid):
    base = dict(n_items=n_i, n_persons=n_p, n_raters=n_r, max_score=m_s,
                item_range=irange, facet_range=frange, person_sd=psd, missing=miss)
    rm = defaultdict(list)

    def take(vec, cvals, fn, gen):
        for c in cvals:
            try:
                mt = recovery_metrics(fn(c), gen)
            except Exception:
                mt = dict(FAIL)
            rm[(vec, c)].append(mt)

    for rep in range(REPS):
        # ---- global representation: item + threshold + global rater stages ----
        sg = rp.MFRM_Sim_Global(n_i, n_p, m_s, n_r, item_range=irange,
                                facet_range=frange, person_sd=psd,
                                missing=miss, seed=next(seeds))
        mg = rp.MFRM(sg.responses, validate=False)
        mg.calibrate(model="global", constant=REF_CONSTANT, method="log-lik")

        def item_fn(c, mg=mg):
            mg.item_diffs(constant=c, method="log-lik")
            return np.asarray(mg.items)
        take("items", CONSTANTS_ITEM, item_fn, np.asarray(sg.items))

        mg.item_diffs(constant=REF_CONSTANT, method="log-lik")          # restore
        ref_items = np.asarray(mg.items)

        def thr_fn(c, mg=mg, ref_items=ref_items):
            return np.asarray(mg.ra_thresholds(pd.Series(ref_items, index=mg.items.index),
                                               constant=c))
        take("thresholds", CONSTANTS_THRESH, thr_fn, np.asarray(sg.thresholds))

        def rg_fn(c, mg=mg):
            mg._estimate_raters_global(constant=c, method="log-lik")
            return np.asarray(mg.facet_effects_global)
        take("rater_global", CONSTANTS_RATER, rg_fn, np.asarray(sg.facet_effects))

        # ---- matrix representation: matrix rater stage ----
        sm = rp.MFRM_Sim_Matrix(n_i, n_p, m_s, n_r, item_range=irange,
                                item_facet_range=frange, threshold_facet_range=frange,
                                person_sd=psd, missing=miss, seed=next(seeds))
        mm = rp.MFRM(sm.responses, validate=False)
        mm.calibrate(model="matrix", constant=REF_CONSTANT, method="log-lik")

        def rm_fn(c, mm=mm):
            mm._estimate_raters_matrix(constant=c, method="log-lik")
            return mm.facet_effects_matrix.values
        take("rater_matrix", CONSTANTS_RATER, rm_fn, np.asarray(sm.facet_effects))

    crows.extend(condition_rows(base, rm))
    if (ci + 1) % max(1, len(grid) // 20) == 0:
        print("  " + _eta(ci + 1, len(grid), t0))

cond = pd.DataFrame(crows)
print("done:", (time.time() - t0) / 60, "min;", len(cond), "condition rows")
````

## Recommended constant per parameter vector

`rmse_opt` minimises mean RMSE; `bias_opt@t` minimises RMSE among constants with |beta - 1| <= t; `minimax` minimises the worst per-condition RMSE ratio vs the best constant for that condition.

````python
rec = recommend(cond)
rec.to_csv(OUTDIR / 'MFRM_recommendation.csv', index=False)
rec
````

## Bias vs RMSE sweet spot

`tradeoff(cond, vector)` lists, per constant, the compression `|beta-1|`, the RMSE and its ratio to the best, whether the point is Pareto-optimal, and a scalarised `score = (rmse/min_rmse - 1) + bias_weight*|beta-1|`. `sweet_spots` reports the score-minimising constant at bias weights 0.5 / 1 / 2. This is the key output for MFRM `rater_matrix`.

````python
print(sweet_spots(cond).to_string(index=False))
for vec in sorted(cond['vector'].unique()):
    print(f'\n===== {vec} =====')
    print(tradeoff(cond, vec).to_string())
sweet_spots(cond).to_csv(OUTDIR / 'MFRM_sweet_spots.csv', index=False)
````

### Breakdown of the RMSE-optimal / bias-optimal constant by key factors

````python
def _picks(g):
    m = _agg(g, ['constant']).set_index('constant')
    rmse_opt = m['rmse'].idxmin() if m['rmse'].notna().any() else np.nan
    if m['beta'].notna().any():
        near = m[(m['beta'] - 1).abs() <= 0.02]
        bias_opt = (near['rmse'].idxmin() if not near.empty
                    else (m['beta'] - 1).abs().idxmin())
    else:
        bias_opt = rmse_opt
    return pd.Series({'rmse_opt': rmse_opt, 'bias_opt': bias_opt})
for fac in [c for c in ['n_items','n_persons','n_raters','max_score','item_range','category_base','facet_range','person_sd','missing'] if c in cond.columns]:
    tbl = cond.groupby(['vector', fac]).apply(_picks, include_groups=False)
    print(f'\n--- by {fac} ---'); print(tbl.to_string())
````

## Plots

````python
for vec in sorted(cond['vector'].unique()):
    facet = 'n_raters' if ('rater' in vec and 'n_raters' in cond.columns) else 'n_items'
    plot_vector(cond, vec, facet=facet, fname='MFRM_' + vec + '.png')
    plot_tradeoff(cond, vec, fname='MFRM_' + vec + '_tradeoff.png')
````

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.stats import gaussian_kde

import raschpy as rp
```

```python
slm_sim = rp.SLM_Sim(no_of_persons=500,
                     no_of_items=30,
                     item_range=2,
                     offset=0.5,
                     seed=999)

slm = rp.SLM(slm_sim)
slm.calibrate()
slm.item_stats_df()
```

```python
slm.wright_map(figsize=(15, 8), map_type='hist', item_labels=True, orientation='horizontal',
               distribution_markers=True, pad=True, prop=False, bw_method='silverman')
```

```python
rsm_sim = rp.RSM_Sim(no_of_persons=500,
                     no_of_items=30,
                     max_score=5,
                     item_range=2,
                     offset=0.5,
                     seed=999)

rsm = rp.RSM(rsm_sim)
rsm.calibrate()
rsm.item_stats_df()
```

```python
rsm.wright_map(map_type='hist', item_labels=True, orientation='horizontal', distribution_markers=True, pad=True, prop=False, bw_method='silverman')
```

```python
import raschpy as rp
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
def rmse(x, y):
    se = (x - y) ** 2
    return (se.mean()) ** 0.5
```

```python
missing_list = [0, 0.25, 0.5, 0.7, 0.9]
n_persons_list = [50, 100, 150, 500, 1000, 10000]
n_items_list = [10, 20, 40, 100, 1000]
```

```python
%%time
seed = 1

results_lol = {missing: {no_of_persons: {no_of_items: {}
                                         for no_of_items in n_items_list}
                         for no_of_persons in n_persons_list}
               for missing in missing_list}

results_cos = {missing: {no_of_persons: {no_of_items: {}
                                         for no_of_items in n_items_list}
                         for no_of_persons in n_persons_list}
               for missing in missing_list}

for missing in missing_list:
    for no_of_persons in n_persons_list:
        for no_of_items in n_items_list:
            for sim in range(20):
                print(f'missing {missing}, n_persons {no_of_persons}, n_items {no_of_items}, sim {sim + 1}')
            
                slm_sim = rp.SLM_Sim(no_of_items = no_of_items,
                                     no_of_persons = no_of_persons,
                                     missing=missing,
                                     item_range=1 + (sim + 1) / 5,
                                     person_sd=max(2, 1 + (sim + 1) / 10))

                slm = rp.SLM(slm_sim)

                slm.calibrate()
                results_lol[missing][no_of_persons][no_of_items][sim] = rmse(slm.items, slm.generating.items)

                slm.calibrate(method='cos')
                results_cos[missing][no_of_persons][no_of_items][sim] = rmse(slm.items, slm.generating.items)

                seed += 1
                                     
```

```python
results_lol = pd.DataFrame.from_dict({(missing, n_p, n_i): sims
                                        for missing, d1 in results_lol.items()
                                        for n_p, d2 in d1.items()
                                        for n_i, sims in d2.items()
                                        }, orient="index").stack().reset_index()

results_lol.columns = ["missing", "no_of_persons", "no_of_items", "sim", "rmse"]
```

```python
results_cos = pd.DataFrame.from_dict({(missing, n_p, n_i): sims
                                        for missing, d1 in results_cos.items()
                                        for n_p, d2 in d1.items()
                                        for n_i, sims in d2.items()
                                        }, orient="index").stack().reset_index()

results_cos.columns = ["missing", "no_of_persons", "no_of_items", "sim", "rmse"]
```

```python
(results_cos['rmse'] < results_lol['rmse']).mean()
```

```python
(results_cos['rmse'] < results_lol['rmse']).median()
```

```python
(results_cos['rmse'] - results_lol['rmse']).mean()
```

```python
import seaborn as sns

sns.kdeplot(results_cos['rmse'] - results_lol['rmse'])
```

```python
plt.scatter(results_cos['rmse'], results_lol['rmse'])
```

```python
for missing in missing_list:
    data_cos = results_cos[results_cos['missing'] == missing]
    data_lol = results_lol[results_lol['missing'] == missing]
    
    plt.scatter(data_cos['rmse'], data_lol['rmse'], label=f'missing={missing}', s=3)

plt.legend()
```

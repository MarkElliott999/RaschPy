```python
# Workaround: this Mac's toolchain has dropped the classic ld64 linker that
# pytensor still asks for on macOS >= 15, so its default C-compile flags fail.
# Compiling without -ld64 works fine here, so strip just that flag rather than
# disabling C compilation (which would fall back to a much slower NumPy backend).
import pytensor.link.c.cmodule as _cmodule

_orig_compile_args = _cmodule.GCC_compiler.compile_args
def _compile_args_no_ld64(march_flags=True):
    return [f for f in _orig_compile_args(march_flags) if f != '-ld64']
_cmodule.GCC_compiler.compile_args = _compile_args_no_ld64
```

```python
import pymc as pm
import numpy as np
import pandas as pd
import arviz as az

from matplotlib import pyplot as plt

import raschpy as rp
```

```python
def recovery_plot(generating, estimated, label, filename=None):
    x, y = np.asarray(generating), np.asarray(estimated)
    fig, ax = plt.subplots()
    ax.scatter(x, y, alpha=0.6)
    lo, hi = min(x.min(), y.min()), max(x.max(), y.max())
    ax.plot([lo, hi], [lo, hi], color='DarkRed', label='Identity')
    m, b = np.polyfit(x, y, 1)
    ax.plot([lo, hi], [m * lo + b, m * hi + b], color='Red', linestyle='--', label='Regression')
    ax.set_xlabel(f'Generating {label}')
    ax.set_ylabel(f'Estimated {label}')
    ax.set_aspect('equal')
    ax.legend()
    if filename is not None:
        plt.savefig(filename)
    plt.show()
    print(f'{label} Pearson correlation:               {round(np.corrcoef(x, y)[0, 1], 3)}')
    print(f'{label} SD ratio (estimated / generating): {round(y.std() / x.std(), 3)}')
    print(f'{label} regression coefficient:            {round(m, 3)}')
    print(f'{label} RMSE:                              {round(np.sqrt(((x - y) ** 2).mean()), 3)}')
```

Create simulated SLM data from known parameters using `RaschPy`

```python
no_of_items = 30
no_of_persons = 100

diffs = np.random.uniform(0, 1, no_of_items)
diffs /= (diffs.max() - diffs.min())
diffs -= diffs.mean()
diffs *= 4

slm_sim = rp.SLM_Sim(no_of_items=no_of_items, no_of_persons=no_of_persons, person_sd=2, offset=0.5,
                     manual_items=diffs, missing=0, seed=42)

slm = rp.SLM(slm_sim)
slm.calibrate()
```

```python
long_data = slm_sim.responses.unstack().dropna()
long_data = long_data.reset_index()
long_data.columns = ['Item', 'Person', 'Response']
long_data
```

```python
person_codes, person_labels = pd.factorize(long_data['Person'])
item_codes, item_labels = pd.factorize(long_data['Item'])

person_idx = person_codes.astype(np.int64)
item_idx = item_codes.astype(np.int64)
observed_responses = long_data['Response'].values

no_of_persons = len(person_labels)
no_of_items = len(item_labels)
```

Define priors

```python
mu_diffs = np.zeros(no_of_items)
#sigma_diffs = np.ones(no_of_items)
sigma_diffs = 5 * np.ones(no_of_items)
```

Build the Bayesian Rasch Model

```python
with pm.Model() as rasch_model:
    abils = pm.Normal('abil', mu=0, sigma=1, shape=no_of_persons)
    diffs = pm.Normal('diff', mu=mu_diffs, sigma=sigma_diffs, shape=no_of_items)
    
    # Deterministic logit link function
    # Probability for each observed combination
    p = pm.math.invlogit(abils[person_idx] - diffs[item_idx])
    
    # Likelihood (Missing data is bypassed by modeling only the observed long-format rows)
    Y_obs = pm.Bernoulli('Y_obs', p=p, observed=observed_responses)
    
    # MCMC Sampling
    idata = pm.sample(draws=1000, tune=1000, return_inferencedata=True, random_seed=42)
```

Sample from Posterior

```python
%%time
with rasch_model:
    idata = pm.sample(draws=1000, tune=1000, return_inferencedata=True)
```

View plot of estimated versus generating item difficulties

```python
estimates_uninformative = az.summary(idata, var_names=['diff'])['mean']
estimates_uninformative -= estimates_uninformative.mean()
estimates_uninformative.index = item_labels
```

Create informative priors based on a given correlation

```python
correlation = 0.6
item_range = 4
```

```python
second_diffs = np.random.normal(0, 1, no_of_items)
second_diffs -= second_diffs.mean()
second_diffs *= slm.generating.items.std()

priors = (correlation * slm.generating.items) + (((1 - correlation ** 2) ** 0.5) * second_diffs)
```

```python
long_data = slm_sim.responses.unstack().dropna()
long_data = long_data.reset_index()
long_data.columns = ['Item', 'Person', 'Response']
long_data
```

```python
person_codes, person_labels = pd.factorize(long_data['Person'])
item_codes, item_labels = pd.factorize(long_data['Item'])

person_idx = person_codes.astype(np.int64)
item_idx = item_codes.astype(np.int64)
observed_responses = long_data['Response'].values

no_of_persons = len(person_labels)
no_of_items = len(item_labels)
```

```python
mu_diffs = priors
sigma_diffs = np.ones(no_of_items)
```

```python
with pm.Model() as rasch_model:
    abils = pm.Normal('abil', mu=0, sigma=1, shape=no_of_persons)
    diffs = pm.Normal('diff', mu=mu_diffs, sigma=sigma_diffs, shape=no_of_items)
    
    # Deterministic logit link function
    # Probability for each observed combination
    p = pm.math.invlogit(abils[person_idx] - diffs[item_idx])
    
    # Likelihood (Missing data is bypassed by modeling only the observed long-format rows)
    Y_obs = pm.Bernoulli('Y_obs', p=p, observed=observed_responses)
    
    # MCMC Sampling
    idata = pm.sample(draws=1000, tune=1000, return_inferencedata=True, random_seed=42)
```

```python
%%time
with rasch_model:
    idata = pm.sample(draws=1000, tune=1000, return_inferencedata=True)
```

```python
estimates_informative = az.summary(idata, var_names=['diff'])['mean']
estimates_informative -= estimates_informative.mean()
estimates_informative.index = item_labels
```

```python
results_df = pd.DataFrame()
results_df['Generating'] = slm.generating.items
results_df['RaschPy'] = slm.items
results_df['Uninformative priors'] = estimates_uninformative
results_df['Priors'] = priors
results_df['Informative priors'] = estimates_informative
```

```python
results_df.round(3)
```

# PCM constant redo: post same-item-cell fix

`PCM._build_pairwise_matrix()` used to let same-item cells leak into the PAIR
comparison matrix -- adjacent-threshold same-item cells got a nonzero entry
that was really just a category-frequency count mislabelled as a comparison,
and non-adjacent same-item cells were a permanent structural zero regardless
of N (a person has one score per item, so two different categories of the
same item can never both be true for them -- PAIR only ever compares pairs of
**distinct** items). Both are now excluded from the matrix entirely.

A direct check found this bug was almost certainly the source of the
"~13-16% outward scale bias (beta~1.13-1.16 at c=0)" that motivated the
entire prior blend / two-pass / 3pB PCM constant-rule investigation -- at
c=0.001 on the fixed matrix, beta is 0.997-1.030 across K=3/5/7, N=800/5000.
That means static-vs-dynamic, one-pass-vs-two-pass are **all** back in play,
not just due for a rerun with the same conclusions.

This notebook also pegs `person_sd` to the combined item+threshold range
(the same deconfounding applied to SLM/PCM/RSM/MFRM's items/thresholds/
stretch-model constants earlier this session), which the original PCM
constant study never controlled for either -- old fixed person_sd=1.5 is kept
as an explicit narrow-SD edge case, not an accidental universal setting.

```
import sys, time, warnings, itertools
from pathlib import Path
import numpy as np, pandas as pd
warnings.filterwarnings("ignore")
import raschpy as rp
print("raschpy:", rp.__file__)

OUT = Path("constant_study_outputs"); OUT.mkdir(exist_ok=True)
RUN_TAG = "postfix"
```

## Deconfounding + recovery helpers

```
def pegged_sd(item_range, K, category_base):
    return max(0.5, (item_range + (K - 1) * category_base) / 4)

NARROW_SD = 1.5   # the OLD fixed default -- deliberate edge case now, not an accident

def resolve_sd(item_range, K, category_base, mode):
    return pegged_sd(item_range, K, category_base) if mode == "pegged" else NARROW_SD

def rec(gen, est):
    x, y = np.asarray(gen, float).ravel(), np.asarray(est, float).ravel()
    if x.shape != y.shape:
        return (np.nan, np.nan)
    ok = np.isfinite(x) & np.isfinite(y)
    x, y = x[ok], y[ok]
    if len(x) < 2:
        return (np.nan, np.nan)
    xc, yc = x - x.mean(), y - y.mean()
    ss = float(xc @ xc)
    rmse = float(np.sqrt(np.mean((xc - yc) ** 2)))
    beta = float(xc @ yc / ss) if (xc.std() >= 1e-6 and ss > 0) else np.nan
    return (rmse, beta)

# WIDE constant grid -- deliberately extended past the old study's ceiling
# (which silently censored its own fit at high co-observation count; see the
# memory's CENSORED-GRID finding). If beta stays >= FLOOR well past c=2.5 now,
# that's real signal, not a grid artefact.
CS = [0.01, 0.02, 0.05, 0.1, 0.2, 0.3, 0.63, 1.0, 1.5, 2.5, 5.0, 10.0, 20.0]
REPS = 30
SEED0 = 31000
FLOOR = 0.85

def item_thr_recovery(m, sim):
    ir, ib = rec(sim.items.values, m.items.values)
    tr, tb = rec(sim.thresholds.values, m.thresholds.values)
    return ir, ib, tr, tb
```

## Manual two-pass rule (prior investigation's construction)

Reused as-is from the prior blend/two-pass/3pB work so the notebook can
answer directly whether it still helps -- NOT refit here. If a flat scalar
turns out to be enough post-fix, this comparison closes the question rather
than reopening the whole fitting exercise.

```
def _solve_from_matrix(m, matrix, constant, method="log-lik", matrix_power=None):
    mp = m._resolve_matrix_power(method, matrix_power)
    cum_scores = np.concatenate(([0], np.cumsum(m.max_score_vector.to_numpy())))
    constant_matrix = ((matrix + matrix.T) > 0).astype(np.float64) * constant
    mat_c = matrix + constant_matrix
    np.fill_diagonal(mat_c, mat_c.diagonal() + constant)
    with np.errstate(divide="ignore", over="ignore", invalid="ignore"):
        if mp == 0:
            mat = np.array(mat_c, dtype=np.float64)
            if constant:
                off = ~np.eye(mat_c.shape[0], dtype=bool)
                mat[off & (mat == 0)] = constant
        else:
            mat = np.linalg.matrix_power(mat_c, mp)
            mat_pow = mp
            while 0 in mat:
                mat = mat @ mat_c
                mat_pow += 1
                if mat_pow == mp + 5:
                    mat += constant
                    break
    tv = m.priority_vector(mat, method=method, pcm=True)
    tv = tv.to_numpy() if hasattr(tv, "to_numpy") else np.array(tv)
    split_indices = cum_scores[1:-1]
    arrays = np.split(tv, split_indices)
    return arrays

def blend_constant(meancount, K, missing):
    # one-pass 2-D rule from the prior investigation (geomean of thr & item
    # optima) -- kept verbatim as the pass-1 base, not refit.
    return float(np.clip(0.0865 * K ** -1.72 * meancount * np.exp(1.27 * missing), 0.02, 60.0))

def two_pass_thresholds(m, K_eff=None):
    matrix, row_items = m._build_pairwise_matrix()
    off = ~np.eye(matrix.shape[0], dtype=bool)
    meancount = float((matrix + matrix.T)[off].mean()) if off.any() else 0.0
    missing = float(m.responses.isna().mean().mean())
    K = K_eff if K_eff is not None else int(m.max_score_vector.max())
    c1 = 2.5 * blend_constant(meancount, K, missing)
    arrays1 = _solve_from_matrix(m, matrix, c1)
    dhat = np.concatenate(arrays1)
    # gap-tapered pass 2
    D = matrix.shape[0]
    gap = np.abs(dhat[:, None] - dhat[None, :])
    h = -max(0.6 / K, 0.1)
    c0 = c1
    c_ab = np.maximum(c0 * (1 + h * gap), 0.02)
    mp = m._resolve_matrix_power("log-lik", None)
    constant_matrix = ((matrix + matrix.T) > 0).astype(np.float64) * c_ab
    mat_c = matrix + constant_matrix
    np.fill_diagonal(mat_c, mat_c.diagonal() + np.diagonal(c_ab))
    with np.errstate(divide="ignore", over="ignore", invalid="ignore"):
        mat = np.array(mat_c, dtype=np.float64)
        off_mask = ~np.eye(mat_c.shape[0], dtype=bool)
        mat[off_mask & (mat == 0)] = c0
    tv = m.priority_vector(mat, method="log-lik", pcm=True)
    tv = tv.to_numpy() if hasattr(tv, "to_numpy") else np.array(tv)
    cum_scores = np.concatenate(([0], np.cumsum(m.max_score_vector.to_numpy())))
    split_indices = cum_scores[1:-1]
    arrays2 = np.split(tv, split_indices)
    meancount_val = meancount
    return arrays1, arrays2, meancount_val

def centred_from_arrays(m, arrays):
    centred = {}
    means = {}
    for i, item in enumerate(m.responses.columns):
        arr = arrays[i]
        centred[item] = arr - arr.mean()
        means[item] = arr.mean()
    max_len = max(len(v) for v in centred.values())
    out = np.full((len(centred), max_len), np.nan)
    for i, item in enumerate(m.responses.columns):
        out[i, :len(centred[item])] = centred[item]
    items_est = np.array([means[c] for c in m.responses.columns])
    return out, items_est
```

## Nested-subset N design

One N=10000 `PCM_Sim` per (condition, rep); smaller N = the first `n_sub`
rows of the SAME dataset, so between-N comparisons aren't contaminated by
fresh sampling noise on top of the effect being measured (reused from the
prior investigation's `pcm_nested.py` design).

```
N_LEVELS = [300, 800, 2000, 5000, 10000]

def nested_subsets(sim, n_levels=N_LEVELS):
    full = sim.responses
    return {n: full.iloc[:n] for n in n_levels}
```

## Grid

Baseline: I=12, item_range=3, category_base=1, person_sd pegged.
- **Core**: K in {2,3,4,5,6,7,8} x N (nested subset) x missing {0, 0.3, 0.6,
  0.85}, pegged vs narrow -- the primary re-derivation grid.
- **Mixed max_score_vector**: half K_lo / half K_hi, a few spreads.
- **category_base x K**: range-vs-raw-K cross, matching the SLM/PCM/RSM/MFRM
  pattern.
- **item_range** sweep, pegged vs narrow.

```
I = 12
BASE_ITEM_RANGE = 3.0
BASE_CAT_BASE = 1.0

CORE_KS = [2, 3, 4, 5, 6, 7, 8]
CORE_MISSING = [0.0, 0.3, 0.6, 0.85]

print(f"{len(CORE_KS)} K values x {len(N_LEVELS)} N levels x {len(CORE_MISSING)} missing "
      f"x 2 sd_modes = {len(CORE_KS)*len(N_LEVELS)*len(CORE_MISSING)*2} core cells")
```

## Core sweep: uniform K x N(nested) x missing, pegged vs narrow

For each (K, missing, sd_mode) condition, one N=10000 sim per rep, nested
subsets give every N level for free from the same draw.

```
rows_core, t0 = [], time.time()
seed = SEED0
n_conds = len(CORE_KS) * len(CORE_MISSING) * 2
ci = 0
for K in CORE_KS:
    for missing in CORE_MISSING:
        for sd_mode in ["pegged", "narrow"]:
            ci += 1
            sd = resolve_sd(BASE_ITEM_RANGE, K, BASE_CAT_BASE, sd_mode)
            for rep in range(REPS):
                seed += 1
                try:
                    sim = rp.PCM_Sim(no_of_items=I, no_of_persons=10000, max_score_vector=[K]*I,
                                     item_range=BASE_ITEM_RANGE, person_sd=sd, missing=missing, seed=seed)
                except Exception:
                    continue
                subsets = nested_subsets(sim)
                for n_sub, resp_sub in subsets.items():
                    try:
                        m = rp.PCM(resp_sub, max_score_vector=[K]*I, validate=False)
                        for c in CS:
                            m.calibrate(constant=c, method="log-lik")
                            ir, ib, tr, tb = item_thr_recovery(m, sim)
                            rows_core.append(dict(K=K, N=n_sub, missing=missing, sd_mode=sd_mode,
                                                   constant=c, item_rmse=ir, item_beta=ib,
                                                   thr_rmse=tr, thr_beta=tb, person_sd=sd,
                                                   label=f"core_K{K}_miss{missing}_{sd_mode}"))
                    except Exception:
                        pass
    el = (time.time() - t0) / 60
    print(f"  core: K={K} done  {el:.1f}m  eta {el/ci*(n_conds-ci):.1f}m", flush=True)

df_core = pd.DataFrame(rows_core)
df_core.to_csv(OUT / f"PCM_core_{RUN_TAG}.csv", index=False)
print("done,", len(df_core), "rows")
```

## Mixed max_score_vector arm

Half the items at K_lo, half at K_hi. person_sd pegged to the mean per-item
combined item+threshold span across the two K's.

```
def pegged_sd_mixed(item_range, K_vector, category_base):
    spans = [item_range + (k - 1) * category_base for k in K_vector]
    return max(0.5, np.mean(spans) / 4)

MIXED_CONDS = [(3, 5), (2, 6), (2, 8), (3, 8)]
MIXED_N = [800, 5000]
MIXED_MISSING = [0.0, 0.3]

rows_mixed, t0 = [], time.time()
seed = SEED0 + 200000
for k_lo, k_hi in MIXED_CONDS:
    k_vec = [k_hi] * (I // 2) + [k_lo] * (I // 2)
    for sd_mode in ["pegged", "narrow"]:
        sd = pegged_sd_mixed(BASE_ITEM_RANGE, k_vec, BASE_CAT_BASE) if sd_mode == "pegged" else NARROW_SD
        for N in MIXED_N:
            for missing in MIXED_MISSING:
                for rep in range(REPS):
                    seed += 1
                    try:
                        sim = rp.PCM_Sim(no_of_items=I, no_of_persons=N, max_score_vector=k_vec,
                                         item_range=BASE_ITEM_RANGE, person_sd=sd, missing=missing, seed=seed)
                        for c in CS:
                            m = rp.PCM(sim)
                            m.calibrate(constant=c, method="log-lik")
                            ir, ib, tr, tb = item_thr_recovery(m, sim)
                            rows_mixed.append(dict(k_lo=k_lo, k_hi=k_hi, N=N, missing=missing, sd_mode=sd_mode,
                                                    constant=c, item_rmse=ir, item_beta=ib,
                                                    thr_rmse=tr, thr_beta=tb, person_sd=sd,
                                                    label=f"mixed_K{k_lo}-{k_hi}_{sd_mode}"))
                    except Exception:
                        pass
    el = (time.time() - t0) / 60
    print(f"  mixed: K=({k_lo},{k_hi}) done  {el:.1f}m", flush=True)

df_mixed = pd.DataFrame(rows_mixed)
df_mixed.to_csv(OUT / f"PCM_mixed_{RUN_TAG}.csv", index=False)
print("done,", len(df_mixed), "rows")
```

## category_base x K arm

```
CATBASE_CONDS = list(itertools.product([0.5, 1.0, 2.0], [3, 6]))
rows_cb, t0 = [], time.time()
seed = SEED0 + 300000
for cb, K in CATBASE_CONDS:
    for sd_mode in ["pegged", "narrow"]:
        sd = resolve_sd(BASE_ITEM_RANGE, K, cb, sd_mode)
        for rep in range(REPS):
            seed += 1
            try:
                sim = rp.PCM_Sim(no_of_items=I, no_of_persons=2000, max_score_vector=[K]*I,
                                 item_range=BASE_ITEM_RANGE, category_base=cb, person_sd=sd,
                                 missing=0.0, seed=seed)
                for c in CS:
                    m = rp.PCM(sim)
                    m.calibrate(constant=c, method="log-lik")
                    ir, ib, tr, tb = item_thr_recovery(m, sim)
                    rows_cb.append(dict(category_base=cb, K=K, sd_mode=sd_mode, constant=c,
                                         item_rmse=ir, item_beta=ib, thr_rmse=tr, thr_beta=tb,
                                         person_sd=sd, label=f"catbase{cb}_K{K}_{sd_mode}"))
            except Exception:
                pass

df_cb = pd.DataFrame(rows_cb)
df_cb.to_csv(OUT / f"PCM_catbase_{RUN_TAG}.csv", index=False)
print("done,", len(df_cb), "rows")
```

## item_range arm

```
rows_ir, t0 = [], time.time()
seed = SEED0 + 400000
K = 5
for ir_val in [1.5, 3.0, 6.0, 9.0]:
    for sd_mode in ["pegged", "narrow"]:
        sd = resolve_sd(ir_val, K, BASE_CAT_BASE, sd_mode)
        for rep in range(REPS):
            seed += 1
            try:
                sim = rp.PCM_Sim(no_of_items=I, no_of_persons=2000, max_score_vector=[K]*I,
                                 item_range=ir_val, person_sd=sd, missing=0.0, seed=seed)
                for c in CS:
                    m = rp.PCM(sim)
                    m.calibrate(constant=c, method="log-lik")
                    ir_, ib, tr, tb = item_thr_recovery(m, sim)
                    rows_ir.append(dict(item_range=ir_val, K=K, sd_mode=sd_mode, constant=c,
                                         item_rmse=ir_, item_beta=ib, thr_rmse=tr, thr_beta=tb,
                                         person_sd=sd, label=f"itemrange{ir_val}_{sd_mode}"))
            except Exception:
                pass

df_ir = pd.DataFrame(rows_ir)
df_ir.to_csv(OUT / f"PCM_itemrange_{RUN_TAG}.csv", index=False)
print("done,", len(df_ir), "rows")
```

## Two-pass vs flat scalar, post-fix

Runs the prior investigation's two-pass rule (unmodified) alongside the best
flat scalar found in the core grid, on the same conditions -- answers
directly whether two-pass still earns its complexity now the same-item bug
is fixed.

```
rows_tp, t0 = [], time.time()
seed = SEED0 + 500000
TP_KS = [2, 3, 5, 7]
TP_N = [300, 800, 2000, 5000]
TP_MISSING = [0.0, 0.3, 0.6, 0.85]

for K in TP_KS:
    sd = pegged_sd(BASE_ITEM_RANGE, K, BASE_CAT_BASE)
    for N in TP_N:
        for missing in TP_MISSING:
            for rep in range(REPS):
                seed += 1
                try:
                    sim = rp.PCM_Sim(no_of_items=I, no_of_persons=N, max_score_vector=[K]*I,
                                     item_range=BASE_ITEM_RANGE, person_sd=sd, missing=missing, seed=seed)
                    m = rp.PCM(sim)
                    arrays1, arrays2, meancount = two_pass_thresholds(m, K_eff=K)
                    thr_1, items_1 = centred_from_arrays(m, arrays1)
                    thr_2, items_2 = centred_from_arrays(m, arrays2)
                    tr1, tb1 = rec(sim.thresholds.values, thr_1)
                    tr2, tb2 = rec(sim.thresholds.values, thr_2)
                    ir1, ib1 = rec(sim.items.values, items_1)
                    ir2, ib2 = rec(sim.items.values, items_2)
                    # best flat scalar comparator: 0.1 and 0.63 (from the other
                    # three models' post-fix experience) -- cheap to include here
                    row = dict(K=K, N=N, missing=missing, meancount=meancount,
                               thr_rmse_pass1=tr1, thr_beta_pass1=tb1,
                               thr_rmse_pass2=tr2, thr_beta_pass2=tb2,
                               item_rmse_pass1=ir1, item_beta_pass1=ib1,
                               item_rmse_pass2=ir2, item_beta_pass2=ib2)
                    for c in [0.1, 0.63]:
                        m.calibrate(constant=c, method="log-lik")
                        ir, ib, tr, tb = item_thr_recovery(m, sim)
                        row[f"thr_rmse_flat{c}"] = tr
                        row[f"thr_beta_flat{c}"] = tb
                        row[f"item_rmse_flat{c}"] = ir
                        row[f"item_beta_flat{c}"] = ib
                    rows_tp.append(row)
                except Exception:
                    pass
    el = (time.time() - t0) / 60
    print(f"  two-pass: K={K} done  {el:.1f}m", flush=True)

df_tp = pd.DataFrame(rows_tp)
df_tp.to_csv(OUT / f"PCM_twopass_vs_flat_{RUN_TAG}.csv", index=False)
print("done,", len(df_tp), "rows")
```

## Results: RMSE-vs-beta balance, core grid

```
def balance_report(df, label_col="label", rmse_col="thr_rmse", beta_col="thr_beta"):
    for lab in df[label_col].unique():
        g = df[df[label_col] == lab]
        t = g.groupby("constant").agg(rmse=(rmse_col, "mean"), beta=(beta_col, "mean"))
        if t["rmse"].isna().all():
            print(f"  {lab:28} -- no valid data (all NaN) -- check for a structural failure at this condition")
            continue
        eligible = t[t.beta >= FLOOR]
        if len(eligible):
            cstar = eligible["rmse"].idxmin()
        elif t["beta"].notna().any():
            cstar = t["beta"].idxmax()
        else:
            cstar = t["rmse"].idxmin()
        print(f"  {lab:28} c*={cstar:<6g} rmse={t.loc[cstar,'rmse']:.4f} beta={t.loc[cstar,'beta']:.3f}"
              f"  (curve: " + " ".join(f"{c:g}:{t.loc[c,'rmse']:.3f}/{t.loc[c,'beta']:.2f}" for c in CS if c in t.index) + ")")

print("=== core grid: threshold recovery ===")
balance_report(df_core, rmse_col="thr_rmse", beta_col="thr_beta")
print()
print("=== core grid: item recovery ===")
balance_report(df_core, rmse_col="item_rmse", beta_col="item_beta")
```

### Confound check: pegged vs narrow, by K (pooled over N, missing)

```
import re
piv = df_core.groupby(["K", "sd_mode", "constant"]).agg(rmse=("thr_rmse", "mean"), beta=("thr_beta", "mean"))

def pick_cstar(row):
    if row["rmse"].isna().all():
        return None
    elig = row[row.beta >= FLOOR]
    if len(elig):
        return elig["rmse"].idxmin()
    return row["beta"].idxmax() if row["beta"].notna().any() else row["rmse"].idxmin()

for K in sorted(df_core.K.unique()):
    try:
        row_p = piv.loc[(K, "pegged")]
        row_n = piv.loc[(K, "narrow")]
    except KeyError:
        print(f"  K={K:>2}  incomplete -- skipped")
        continue
    cp, cn = pick_cstar(row_p), pick_cstar(row_n)
    if cp is None or cn is None:
        print(f"  K={K:>2}  -- no valid data in {'pegged' if cp is None else 'narrow'} -- skipped")
        continue
    print(f"  K={K:>2}  PEGGED c*={cp:<7g} beta={row_p.loc[cp,'beta']:.3f}  |  "
          f"NARROW c*={cn:<7g} beta={row_n.loc[cn,'beta']:.3f}")
```

### Items-vs-thresholds tension check

The prior two-pass investigation's core rationale was that items and
thresholds wanted genuinely different constants -- especially at K=2, where
"widths want ~5x the c the locations want". Checking directly whether that
tension survives the fix, or was itself downstream of the same-item bias
(which would explain a spurious pull in opposite directions: the bias
inflated threshold spread specifically, so a bigger constant would have
looked like it was needed there while doing nothing for -- or actively
harming -- items sharing the same knob).

```
piv_thr = df_core.groupby(["K", "sd_mode", "constant"]).agg(rmse=("thr_rmse","mean"), beta=("thr_beta","mean"))
piv_item = df_core.groupby(["K", "sd_mode", "constant"]).agg(rmse=("item_rmse","mean"), beta=("item_beta","mean"))

print(f"{'K':>3} {'mode':>7}   {'items c*':>9} {'items beta':>11}   {'thr c*':>8} {'thr beta':>9}   {'ratio (thr/items)':>18}")
for K in sorted(df_core.K.unique()):
    for mode in ["pegged", "narrow"]:
        try:
            row_t = piv_thr.loc[(K, mode)]
            row_i = piv_item.loc[(K, mode)]
        except KeyError:
            print(f"{K:>3} {mode:>7}   incomplete -- skipped")
            continue
        ct, ci = pick_cstar(row_t), pick_cstar(row_i)
        if ct is None or ci is None:
            print(f"{K:>3} {mode:>7}   no valid data -- skipped")
            continue
        ratio = ct / ci if ci else float('nan')
        print(f"{K:>3} {mode:>7}   {ci:>9g} {row_i.loc[ci,'beta']:>11.3f}   {ct:>8g} {row_t.loc[ct,'beta']:>9.3f}   {ratio:>18.2f}")
```

### N dependence, pegged core (does c* still scale with N post-fix?)

```
piv_n = df_core[df_core.sd_mode == "pegged"].groupby(["K", "N", "constant"]).agg(
    rmse=("thr_rmse", "mean"), beta=("thr_beta", "mean"))
for K in sorted(df_core.K.unique()):
    print(f"  K={K}:")
    for N in N_LEVELS:
        try:
            row = piv_n.loc[(K, N)]
        except KeyError:
            print(f"    N={N:>6}  incomplete -- skipped")
            continue
        cstar = pick_cstar(row)
        if cstar is None:
            print(f"    N={N:>6}  -- no valid data")
            continue
        print(f"    N={N:>6}  c*={cstar:<7g} rmse={row.loc[cstar,'rmse']:.4f} beta={row.loc[cstar,'beta']:.3f}")
```

## Results: two-pass vs flat scalar

The key question this section answers: now that the same-item bug is fixed,
does the two-pass taper still beat a flat scalar, or has the gap it was
built to close mostly disappeared?

```
if len(df_tp):
    g = df_tp.groupby(["K", "N", "missing"]).agg(
        pass1=("thr_rmse_pass1", "mean"), pass2=("thr_rmse_pass2", "mean"),
        flat01=("thr_rmse_flat0.1", "mean"), flat063=("thr_rmse_flat0.63", "mean"),
        beta_pass1=("thr_beta_pass1", "mean"), beta_pass2=("thr_beta_pass2", "mean"),
        beta_flat01=("thr_beta_flat0.1", "mean"), beta_flat063=("thr_beta_flat0.63", "mean"),
    )
    print(f"{'K':>3} {'N':>6} {'miss':>5}   {'pass1':>8} {'pass2':>8} {'flat0.1':>8} {'flat0.63':>8}   "
          f"{'b_p1':>6} {'b_p2':>6} {'b_f.1':>6} {'b_f.63':>6}")
    for (K, N, missing), row in g.iterrows():
        print(f"{K:>3} {N:>6} {missing:>5.2f}   {row.pass1:>8.4f} {row.pass2:>8.4f} "
              f"{row.flat01:>8.4f} {row.flat063:>8.4f}   "
              f"{row.beta_pass1:>6.3f} {row.beta_pass2:>6.3f} {row.beta_flat01:>6.3f} {row.beta_flat063:>6.3f}")
else:
    print("no two-pass rows -- check for structural failures upstream")
```

## Notes

- If the core grid's `c*` now sits in a narrow, K-INDEPENDENT band (unlike
  the old K^-1.7 to K^-2.5 finding), the whole dynamic-rule apparatus can
  likely be retired in favour of a flat scalar -- check this first.
- The items-vs-thresholds tension check is the other half of that question:
  if `items c* ~= thresholds c*` post-fix (ratio near 1), the whole reason a
  single scalar `constant` couldn't serve both vectors may have been the
  same-item bias inflating threshold spread specifically -- fix the cause,
  and the two vectors may simply want the same thing again, the way they do
  in SLM/RSM. If the ratio is still far from 1 (especially at K=2, the old
  investigation's worst case), the tension is real and independent of the bug.
- The N-dependence table is the direct test of whether `c*` still needs to
  scale with co-observation count (old finding: c* ~ coobs^0.9, driven by the
  bias this fix removes) -- if `c*` is now roughly flat across N at fixed K,
  that confirms the old scaling was compensating for the bug, not a genuine
  co-observation effect.
- The `sd_mode` confound check separates genuine K-effects from the
  person_sd/range mismatch the original PCM study never controlled for.
- The two-pass-vs-flat-scalar section is the practical bottom line: if a flat
  scalar (or a simple K-only rule, if K-dependence survives) matches two-pass
  within noise, ship the simple version and retire the tapered construction.
- CS extends to 20 specifically to avoid re-censoring the fit the way the
  original grid (capped at 2.0) did -- watch for any condition where beta is
  STILL >= FLOOR at the top of the grid, which would mean the grid needs
  extending further still.
